```python
import cobra

model = cobra.io.read_sbml_model("../models/model_5.xml")  # your latest saved model

# 1. Which compartment does each ATPM metabolite actually belong to?
atpm = model.reactions.get_by_id("ATPM")
for met in atpm.metabolites:
    print(met.id, "-> compartment:", met.compartment)

# 2. What's the most common compartment in the whole model? (this is what
#    MEMOTE assumes is "the main compartment" if it can't find true cytosol)
from collections import Counter
counts = Counter(met.compartment for met in model.metabolites)
print("\nCompartment metabolite counts:", counts)

# 3. Are there any metabolites incorrectly sitting in the stray 'c' compartment?
stray = [met.id for met in model.metabolites if met.compartment == "c"]
print(f"\n{len(stray)} metabolites in the stray 'c' compartment:")
print(stray[:20])
```

```python
for met_id in ["atp_c", "h2o_c", "adp_c", "pi_c", "h_c"]:
    met = model.metabolites.get_by_id(met_id)
    if met.compartment != "C_c":
        met.compartment = "C_c"

cobra.io.write_sbml_model(model, "../models/model_8.xml")
```

```python
for met_id in ["atp_c", "h2o_c", "adp_c", "pi_c", "h_c"]:
    met = model.metabolites.get_by_id(met_id)
    print(met.id, met.formula, met.charge)
```

```python
import cobra

model = cobra.io.read_sbml_model("../models/model_5.xml")
atpm = model.reactions.get_by_id("NTP1")

print("Reaction string:", atpm.reaction)
print("Bounds:", atpm.bounds)
print("Reversibility:", atpm.reversibility)

print("\nStoichiometric coefficients:")
for met, coef in atpm.metabolites.items():
    print(f"  {met.id}: {coef}")

print("\nGPR (gene rule):", repr(atpm.gene_reaction_rule))
print("Genes:", list(atpm.genes))

print("\nSubsystem:", atpm.subsystem)
print("Reaction ID:", atpm.id)
print("Reaction name:", atpm.name)
```
