```python
# import libraries
import cobra
from cobra.flux_analysis import flux_variability_analysis

# import model
model = cobra.io.read_sbml_model("../models/model_4.xml")
```

```python
# duplicate reactions to block (from memote report)
duplicates_to_block = ["ACS_1", "ADPRDP_1", "AMANK_1", "AMID_1", "THPAT", "ASPCT_2", "CBPS_1", "TRPAS1", "CYSS_2", "DADNK_1", "DGNSK_1", "DURIPP_1", "ECOAH1_1", "ECOAH5_1", "GLCNt2r", "GTPDPK_1", "HACD1_1", "HISTP_1", "HMGL_2", "IZPN_1", "MCTC_1", "METSR_S2", "METSabc", "NADK_1", "NTPP1_1", "NTPP10_1", "NTPP11_1", "NTPP2_1", "NTPP8_1", "NTPP9_1", "OCBT_1", "PGLYCP_1", "PNTK_1", "PRFGS_1", "RBFSa_1", "SHSL2r", "THRS_2", "THZPSN_1", "EX_glcn_e", "EX_metsox_S__L_e"]

# Identify reactions that actually exist in the model
reactions_to_remove = [rxn_id for rxn_id in duplicates_to_block if rxn_id in model.reactions]

# Remove the reactions
model.remove_reactions(reactions_to_remove)
```

```python
# unbound flux bounds (from memote report)
unbounded_rxns = ["13PPDH","13PPDH2","4ABUTD","5DGLCNR","5DKGR","ACKr","ACONT","ACONTa","ACONTb","ACTD","ACTD_1","ACTDa","ADAPAT","AH6PI","ALATA_D","ALATA_L","ALAtex","ALCD19","ALCD19y","ALCD2ir","ALCD2x","ALCD2y","ASPTA","ASPtpp","BUTKr","CO2t","CO2tex","CO2tpp","COBALT2tex","COBALTt5","DAPDA","DURIPP","DURIPP_1","ECOAH1","ECOAH1_1","ECOAH3","ETOHtex","ETOHtrpp","FE2tex","FE3PYOVDL2","FFSD1r","G6PDH2r","G6PI","GALKr","GALpts","GAPD","GDH1","GK1","GK2","GLUDxi","GLUDy","GLUDy_1","GLUR","GLYBtex","GLYCtex","GLYCtpp","GLYOX_1","GTHOr","H2Ot","H2Otpp","HACD1","HACD1_1","HACD1i","HACD3","HACD8","HACD8i","HG2tex","HSDxi","HSTPTr","Htex","INOSR","Kt1","LEUTA","LEUtex","LGTHL","LLEUDr","METB1","MNtex","MTAM_1","MTHFC","MTHFD","MYCTR","NADDPp_1","NADKd","NADTRHD","NDPK8","NH3c","NH4t","NH4tex","NH4tpp","OCBT_1","ORNTA","P5CR","P5CRx","PGK","PGM","PGMT","PGMT_2","PGMT_B","PRO1x","PTHPS_1","PYNP1","PYNP1_1","PYNP2r","RECOAH3","RHACOAR80","RU5PS","SDPTA","SHSL2","SHSL2r","SMIA1","SMIB1","SSCOARy","SUCOAS","TARTRtpp","THPAT","UAGPT2_1","VALDHr","VALTA","VALtex","VPAMTr","YUMPS","r2465_1","EX_23dhbzs3_e","EX_4abut_e","EX_cit_e","EX_cmcbtt_e","EX_cpgn_un_e","EX_enter_e","EX_fe2_e","EX_fe3_e","EX_fe3pyovd_kt_e","EX_feoxam_un_e","EX_glyald_e","EX_gthrd_e","EX_h2o_e","EX_h_e","EX_istfrnA_e","EX_istfrnB_e","EX_mcbtt_e","EX_nmn_e","EX_o2_e","EX_salchs2_e","EX_salchs4_e","EX_stfrnA_e","EX_stfrnB_e"]

# Filter out reactions starting with 'EX_'
internal_unbounded = [rxn for rxn in unbounded_rxns if not rxn.startswith("EX_")]

print(f"Total internal unbounded reactions: {len(internal_unbounded)}")
internal_unbounded
```

```python
# Identify reactions that loop infinitely even when starved
## All exchanges closed, no maintenance, objective = 0
## Done inside a context manager so these constraints are NOT saved
## this step is purely diagnostic.

print("\n--- Sealing model & relaxing constraints to find internal loops ---")
internal_loops = None
with model:
    for rxn in model.exchanges:
        rxn.lower_bound = 0
    print(f"Closed {len(model.exchanges)} exchange reactions. Model is now starved.")

    for rxn in model.reactions:
        if "ATPM" in rxn.id.upper() or "MAINT" in rxn.id.upper():
            rxn.lower_bound = 0
            print(f"Relaxed maintenance constraint for: {rxn.id}")

    model.objective = model.problem.Objective(0)

    try:
        fva_closed = flux_variability_analysis(model, fraction_of_optimum=0)
        threshold = 999
        unbounded_mask = (fva_closed['maximum'] >= threshold) | (fva_closed['minimum'] <= -threshold)
        internal_loops = fva_closed[unbounded_mask]
        print(f"Found {len(internal_loops)} reactions that loop infinitely even without food.")
        if len(internal_loops) > 0:
            print("\nTop 15 looping reactions to investigate:")
            print(internal_loops)
    except Exception as e:
        print(f"FVA failed: {e}")
        print("There is still a hardcoded forced-flux reaction preventing the model from resting at 0.")
```

```python
# Create a comprehensive list of the exact reactions to deactivate (0, 0)
internal_duplicates_to_block = internal_loops.index.tolist()

# Apply the block to your model
blocked_count = 0
for rxn_id in internal_duplicates_to_block:
    if rxn_id in model.reactions:
        model.reactions.get_by_id(rxn_id).bounds = (0, 0)
        blocked_count += 1

print(f"Successfully deactivated {blocked_count} internal looping reactions.")
```

```python
# Identify transporters
for rxn_id in internal_unbounded:
    # Safe lookup: skip if the ID is missing from the model
    if rxn_id not in model.reactions:
        continue
        
    rxn = model.reactions.get_by_id(rxn_id)
    
    # Rule: Force unidirectionality on transporters (ending in t, tex, tpp)
    if any(rxn_id.endswith(s) for s in ['t', 'tex', 'tpp']):
        rxn.lower_bound = 0

    # Rule: Remove known duplicate suffixes (e.g., _1, _B)
    elif "_1" in rxn_id or "_B" in rxn_id:
        base_id = rxn_id.replace("_1", "").replace("_B", "")
        # Safe containment check using the .reactions DictList
        if base_id in model.reactions:
            rxn.bounds = (0, 0)
```

```python
# remote unused mets
cobra.manipulation.prune_unused_metabolites(model)
```

```python
# export model
cobra.io.write_sbml_model(model, "../models/model_5.xml")
print(f"exported model")
```
